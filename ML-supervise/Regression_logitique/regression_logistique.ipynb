### Regresion logistique : probabilité qu'un évènement se produise en fonction d'une variable explicative

La régression logistique c'est comprendre comment un modèle transforme des informations sur un client en probabilité  

Problématique : prédire un oui ou un non  ⇒ toujours sur une variable binaire  
(VS régression linéaire sur variable numérique)  
La méthode : modèle apprend sur des cas passés donc on connaît la réponse  

Le résultat : une probabilité entre 0 et 100 % pour chaque nouveau cas
⇒ une équation de régression logistique que l'on peut lire, coeeficient par coefficient.
(ce qui fait monter ou baisser cette proba)

#### Vocabulaire

- Cible : ce que l'on veut prédire ➜ 1 par défaut ou 0   

- Variable explicative : une information connue sur chaque cas  

- Probabilité (p) : la chance que la cible = 1, **toujours** entre 0 et 100 %  

- Cote (odds) : $\frac{\text{chances pour (que l'évènement se produise)}}{\text{chances contre (que l'évènement se produise)}} = \frac{p}{(1 - p)}$  
>*ex.* : 30 % contre 70 % = cote de 0,43  
La cote (VS la proba) va de 0 à l'infini : c'est elle que le modèle fait varier, par multiplication  
➜ une cote de 1 : autant de chance pour que contre (50 / 50 %)  
➜ une cote de 4 : 4 événements pour 1 non-événement (p = 80 %)  
➜ une cote de 0,25  : 1  événements pour 4 non-événement (p = 20 %)  

- Score : résultat de l'équation avant de passer en probabilité (p)  
$p = \frac{1}{(1 + \text{ exp(- score)})}$  
(exp = exponentiel)  
**toujours** : score 0 = 50 %,  
> un score négatif (à gauche de zéro sur l'axe des abscisses) = proba faible  
VS score positif = proba forte  
*ex.* : -3 ➜ 5 % ; -2 ➜ 12 % VS 2 ➜ 88 % ; 3 ➜ 95 %  

- Coefficient : le poids d'une variable dans le score  

- Seuil : proba à partir de laquelle on alerte  

⇒ apprentissage supervisé : la cible doit etre connue pour tous les cas passés.  

Puisque toujours entre 0 et 100 %, représentation par une droite impossible ➜ courbe en S.

$$
\begin{align*}
c = \frac{p}{1-p} & \Leftrightarrow (1 - p)c = \frac{p}{1-p} \times 1 - p \\
&\Leftrightarrow (1 - p)c = p \\
&\Leftrightarrow c - cp = p \\
&\Leftrightarrow c = p + cp \\
&\Leftrightarrow c = p(1 + c) \\
&\Leftrightarrow \frac{c}{1 + c} = p \\
\\
&\quad\boxed{p = \frac{c}{1 + c}}
\end{align*}
$$


#### VS régression linéaire

| | Régression linéaire | Régression logistique |
|---|---|---|
| On prédit | Un nombre : un prix, un chiffre d'affaires | Un oui ou un non, avec sa probabilité |
| Équation | $y = b_0 + b_1 \times 1 + ... + b_k \times k$  | $ \text{score} = b_0 + b_1 \times variable_1 + ... + b_k \times variable_k$ (puis courbe en S) |
| Forme | Une droite, sans limite | Une courbe en S, entre 0 et 100 % |
| Recherche des coefficients | Moindres carrés : on réduit les écarts à la droite | Maximum de vraisemblance : on rend les cas observés les plus probables |
| Lecture d'un coefficient | $+1$ sur $x$ ajoute $b$ à $y$ | $+1$ sur $x$ multiplie la cote par $exp(b)$ |
| Qualité du modèle | R², erreur moyenne | AUC, rappel, précision |

On peut examiner le signe des coefficients, leurs valeurs p et les modalités de référence, mais leur interprétation diffère : en régression linéaire, les coefficients portent sur la valeur moyenne de la cible ; en régression logistique, ils portent sur le logarithme des cotes de l’événement.  
Seule la cible change, donc la forme de la réponse.

### Équation du modèle

$$\text{score} \to \text{proba} \to \text{décision} $$

La courbe en S transforme le score en probabilité  
On compare la probabilité à un seuil : au-dessus, on alerte

#### Le score

On additionne une constante $b_0$ et chaque variable multipliée par son coefficient $b_k \times variable_k$

$$ \text{score} = b_0 + b_1 \times variable_1 + b_2 \times variable_2 + \cdots + b_k \times variable_k$$

$b_0$, la constante est le score d'un client de référence quand toutes les variables valent 0  
$b_k \times variable_k$, le poids de chaque variable ➜ c'est ce que le modèle apprend

#### Trouver les coefficients (estimation à partir des données d'entraînement)

$$\text{coeff. quelconque} \to \text{calcul de la proba de chaque individu} \to \text{comparaison} \to \text{correction} $$

Les coeff. servent à construire la prédiction

comparaison : des proba des évènements  
correction : on ajoute et on recommence jusqu'à avoir des coefficients qui ne bougent plus  
⇒ méthode du maximum de vraisemblance : on garde kes coeff. qui rendent les cas observés les plus probables.  

**Odds ratio** (OR) ou rapport de cotes : compare les cotes d’un événement entre deux situations.  
Pour un coefficient $\beta$, il vaut : $OR = e^\beta$  

*ex.* : $e^3 \approx 20.09$  
($e$ est bien un nombre ! définit par $ln(e)=1$)

#### Lire un coeff. // Odds ratio (OR)

| Coefficient | Odds ratio = exp(coefficient) | Lecture |
|---|---|---| 
| Positif | Supérieur à 1 | La variable aggrave le risque |
| Nul | Égal à 1 | La variable n'a aucun effet |
| Négatif | Inférieur à 1 | La variable protège |

---

*Exemple* | Coefficient | Odds ratio | Lecture
|---|---|---|---| 
| Un incident de paiement de plus | + 1,03 | $e^{1.03} = 2.8$ | La cote de l'évènement est multipliée par 2,8 |
| Client propriétaire | - 1,40 | $e^{- 1.4} = 0.25$ | La cote de l'évènement est divisée par 4 |




À variables comparables :  
- OR = 1 : les cotes ne changent pas ;  
- OR = 2 : les cotes de l’événement sont multipliées par 2 ;   
- OR = 0,5 : elles sont divisées par 2.  

L'odds ratio multiplie la cote, pas la probabilité !  
Un odds ratio de 2 ne veut pas dire « deux fois plus de risque », ni que « la probabilité double » : il veut dire « une cote deux fois plus forte »

> *Par exemple*, si la probabilité initiale est de 20 %, les cotes valent $\frac{0.2}{1 - 0.2} = \frac{0.2}{0.8} = 0.25$  
> Si elles doublent et passent à $0.5$, la probabilité : $$ p = \frac{0.5}{1 + 0.5} \approx 33\% \neq 40\% $$

Pour une variable quantitative continue, l’OR correspond généralement à une augmentation d’une unité de cette variable.  
Pour une variable qualitative nominale (catégorielle), il compare une modalité à sa modalité de référence.

#### La modalité de référence

Chaque coefficient se lit par rapport à la modalité de référence choisit, c'est un point de comparaison.
Donc choisir une autre référence changerait l'interprétation du coefficient (pas les prédictions du modéle).

On choisit souvent comme référence la modalité la plus fréquente, celle dont la lecture est plus naturelle.

> *Ex.* : CDD : + 1,85 ⇒ par rapport à la modalité de réf. CDI, les cotes de l'évènement defaut (compare les chances qu'un defaut survienne) associées au CDD sont multipliées par $e^{1.85} \approx 6.3$

On dit : 
- la cote lorsqu’on parle d’un seul profil ou groupe
- les cotes lorsqu’on compare deux profils ou groupes — par exemple CDI et CDD. L’odds ratio de 6,3 signifie alors que la cote de défaut du profil CDD est 6,3 fois la cote de défaut du profil CDI, toutes choses égales par ailleurs.


#### Valeur $p$ et intervalle de confiance

**Valeur $p$** répond à la question : cet effet peut-il venir du hasard ? 
⇒ p < 5 % on considère comme l'effet établi

On teste généralement l’hypothèse qu’il n’y a pas d’association : l’odds ratio vaut 1. Autrement dit, les cotes sont les mêmes entre les deux groupes.  

La valeur p mesure à quel point les données observées seraient surprenantes si cette absence d’association était vraie.  

Si $p < 0.05$, les données sont considérées comme suffisamment incompatibles avec l’hypothèse « pas d’association » pour qu’on parle d’un résultat statistiquement significatif.
Si $p \geq 0.05$, les données ne permettent pas d’écarter cette hypothèse.  

Attention : une valeur p de 0,03 ne signifie pas qu’il y a 3 % de chances que l’effet soit dû au hasard.  
Et $p < 0.05$ ne prouve pas que l’effet existe : c’est un seuil de décision conventionnel, pas une certitude.  
« Non établi » ne veut pas dire « sans effet » : les données ne permettent pas de conclure.

**Intervalle de confiance** est la fourchette dans laquelle se trouve probablement l'OR.
⇒ s'il contient 1, l'effet n'est pas établi

L’intervalle de confiance à 95 % donne une plage de valeurs compatibles avec les données et le modèle.  
Il indique aussi la précision de l’estimation : un intervalle étroit est plus précis, un intervalle large l’est moins.

Pour un odds ratio :

- OR = 1 correspond à l’absence de différence entre les groupes  
- si l’intervalle contient 1, le résultat n’est généralement pas significatif au seuil de 5 %  
- s’il ne contient pas 1, il est généralement significatif à ce seuil

Par exemple, un OR estimé à 2,8 avec un intervalle $[1.4 ; 5.0]$ suggère une association positive, statistiquement significative, mais dont l’ampleur reste incertaine.  
Avec $[0.8 ; 4.2]$, l’intervalle contient 1 : les données ne permettent pas de conclure clairement à une différence.

*Cette correspondance entre l’intervalle et le seuil de 5 % vaut généralement pour un intervalle à 95 % et un test bilatéral construit de façon cohérente.*

> *Ex.* : Le modèle estime un odds ratio de 2,8.  
Pour savoir si cette différence pourrait s’expliquer par les fluctuations des données, on regarde la valeur p et l’intervalle de confiance.

Exemple 1 : $p = 0.03$, intervalle de confiance $[1.2 ; 6.5]$.
Si, en réalité, il n’y avait aucune différence entre CDD et CDI, obtenir par hasard un résultat aussi marqué serait assez rare — environ 3 % dans les conditions du test. L’intervalle ne contient pas 1, ce qui va aussi dans le sens d’une différence. Mais son ampleur reste incertaine : l’OR pourrait être autour de 1,2 comme de 6,5.

Exemple 2 : $p = 0.20$, intervalle $[0.7 ; 5.4]$.
Les données sont trop imprécises pour écarter clairement l’absence de différence : l’intervalle contient 1. Cela ne prouve pas qu’il n’y a pas de différence ; il manque peut-être simplement des données pour la mettre en évidence.

⇒ la valeur p indique si le résultat serait étonnant en l’absence de différence ; l’intervalle indique quelles ampleurs de différence restent compatibles avec les données.  

#### Conditions pour utiliser la méthode

|Condition|Pourquoi|Comment vérifier|
|---|---|---|
|**Une cible à deux valeurs**|Le modèle prédit une probabilité de 1|La cible ne contient que 0 et 1|
|**Un effet dans un seul sen**s|Le coefficient dit « monte » ou « baisse », pas les deux|Le taux de défaut par tranche monte ou baisse régulièrement|
|**Des variables qui ne se répètent pas**|Deux variables liées brouillent leurs coefficients|Le VIF de chaque variable reste sous 5|
|**Assez de cas**|Peu de défauts : coefficients instables|Au moins 10 défauts par coefficient|
|**Des variables à la même échelle**|Pour comparer les coefficients entre eux|On centre et on réduit les variables numériques|


Ces vérifications se font avant d'estimer le modèle. Un coefficient surprenant, de signe contraire au bon sens, vient souvent de deux variables qui se répètent.

VIF signifie Variance Inflation Factor, facteur d’inflation de la variance.  
Par exemple, si deux variables apportent presque la même information, il devient difficile pour le modèle de séparer leur contribution. Leurs coefficients peuvent alors devenir instables et plus difficiles à interpréter.

- VIF proche de 1 : peu de lien avec les autres variables explicatives   
- VIF supérieur à 5 : possible problème de redondance à examiner   
- VIF très élevé : les variables sont probablement fortement redondantes

Le seuil de 5 est une règle pratique, pas une frontière absolue. Un VIF élevé ne veut pas forcément dire qu’il faut supprimer la variable : il invite à examiner les variables concernées. Le VIF renseigne sur les liens entre variables explicatives, pas sur leur lien avec la cible.

#### Le seuil (du score à la décision)

|Mesure|Calcul|Question posée|
|---|---|---|
|Rappel|défauts détectés / défauts réels|Combien de défauts ai-je vus ?|
|Précision|défauts détectés / alertes|Mes alertes sont-elles justifiées ?|

On choisit le seuil selon le coût des deux erreurs. Si un défaut manqué coûte plus cher qu'une fausse alerte, on baisse le seuil.

#### AUC et tranches de score : mesurer la qualité 

Ces mesures se calculent sur des clients que le modèle n'a jamais vus : le jeu de test !

**AUC** : le modèle classe-t-il bien ?   
Le modèle attribue à chaque client un score de risque. On prend au hasard un client qui fera défaut et un qui ne fera pas défaut : l’AUC mesure la probabilité que le modèle donne un score plus élevé au client qui fera défaut.
- AUC = 0,5 : le classement ne fait pas mieux que le hasard  
- AUC = 1 : le modèle classe parfaitement les deux catégories

*Ex.* : une AUC de 0,80 signifie que, sur ce type de paire choisie au hasard, le score du client qui fera défaut est plus élevé dans environ 80 % des cas. Cela ne veut pas dire que 80 % des clients seront correctement prédits !

**Les tranches** : le score dit-il vrai ? (les probabilités prédites correspondent aux fréquences observées)   
On regroupe les clients selon le risque prédit. Parmi ceux à qui le modèle attribue environ 30 % de risque, on vérifie quelle proportion a réellement fait défaut.

*Ex.* : si le groupe contient 100 clients, si les probabilités sont bien calibrées, on s’attend à ce qu’environ 30 fassent défaut (il peut y avoir des variations, surtout dans les petits groupes).

⇒ Un modèle peut bien classer les clients tout en annonçant des probabilités mal calibrées

#### Forces et limites (VS d'autres méthodes)

|Forces|Limites|
|---|---|
|Chaque coeff. se lit et s'explique|Elle suppose un effet régulier de chaque variable|  
|Elle donne une vraie probabilité|Elle ne trouve pas seule les combinaisons de variables|
|Rapide, stable, peu de réglages|Elle demande une préparation des données|  


##### En comparaison : 

|Méthode|Lisibilité|Préparation des données|Quand la choisir|
|---|---|---|---|
|Régression logistique|Forte : une équation|Colonnes 0 / 1, standardisation|Il faut expliquer et justifier la décision|
|Arbre de décision|Forte : des règles|Presque aucune|On veut des règles simples à appliquer|
|Forêt aléatoire, CatBoost|Faible|Presque aucune|Seule la qualité de prédiction compte|
