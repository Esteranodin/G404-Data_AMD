### Comprendre comment un modèle transforme des informations sur un client en probabilité  

Problématique : prédire un oui ou un non  ⇒ toujours sur une variable binaire  
(VS régression linéaire sur variable numérique)  
La méthode : modèle apprend sur des cas passés donc on connaît la réponse  

Le résultat : une probabilité entre 0 et 100 % pour chaque nouveau cas
⇒ une équation de régression logistique que l'on peut lire, coeeficient par coefficient.
(ce qui fait monter ou baisser cette proba)

#### Vocabulaire

- Cible : ce que l'on veut prédire ➜ 1 par défaut ou 0   

- Variable explicative : une information connue sur chaque cas  

- Probabilité (p) : la chance que la cible = 1, entre 0 et 100 %  

- Cote : $\frac{\text{chances pour}}{\text{chances contre}} = \frac{p}{(1 - p)}$  
*ex.* : 30 % contre 70 % = cote de 0,43
➜ une cote de un : autant de chance pour que contre (50 / 50 %)

- Score : résultat de l'équation avant de passer en probabilité (p)  
$p = \frac{1}{(1 + \text{ exp(- score)})}$  
(exp = exponentiel)
**toujours** : score 0 = 50 %,  
un score négatif (à gauche de zéro sur l'axe des abscisses) = proba faible  
VS score positif = proba forte
*ex.* : -3 ➜ 5 % ; -2 ➜ 12 % VS 2 ➜ 88 % ; 3 ➜ 95 %  

- Coefficient : le poids d'une variable dans le score  

- Seuil : proba à partir de laquelle on alerte  

⇒ apprentissage supervisé : la cible doit etre connue pour tous les cas passés.  

Puisque toujours entre 0 et 100 %, représentation par une droite impossible ➜ courbe en S.
